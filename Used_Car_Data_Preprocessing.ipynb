```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import(
    OneHotEncoder,
    OrdinalEncoder,
    StandardScaler,
    MinMaxScaler
)
from sklearn.compose import ColumnTransformer
```

```python
file_name = "Day12_Used_Car_Preprocessing_Dataset.csv"
car_data = pd.read_csv(file_name)
print("Dataset loaded successfully.")
```

```python
print("========== DATASET SHAPE ==========")
print(f"Rows: {car_data.shape[0]}")
print(f"Columns: {car_data.shape[1]}")
print(f"Shape: {car_data.shape}")
```

```python
print("========== COLUMN NAMES ==========")
for column_name in car_data.columns:
  print("-", column_name)
```

```python
print("========== DATA TYPES ==========")
print(car_data.dtypes)
```

```python
print("========== DATASET INFORMATION ==========")
car_data.info()
```

```python
print("========== DESCRIPTIVE STATISTICS ==========")
display(car_data.describe())
```

```python
print("========== MISSING VALUES ==========")
missing_values = car_data.isnull().sum()
display(missing_values)
total_missing = missing_values.sum()
print(
    f"Total missing values: {total_missing}"
)
```

```python
print("========== DUPLICATE RECORDS ==========")
duplicate_count = car_data.duplicated().sum()
print(
    f"Number of duplicate rows: {duplicate_count}"
)
```

```python
if duplicate_count > 0:
  car_data = car_data.drop_duplicates()

  print(
      f"Duplicates removed successfully."
  )
else:
  print("No duplicate records found.")
```

```python
target_column = "Resale_Price_Lakh"
print(
    f"Target variable: {target_column}"
)
```

```python
car_data = car_data.drop(
    columns=["Car_ID"]
)
print(
    "Car_ID removed because it is an identifier"
)
```

```python
x = car_data.drop(
    columns=[target_column]
)
y = car_data[target_column]
print("========== FEATURES ==========")
print(x.columns.tolist())
print("\n========== TARGET ==========")
print(y.name)
```

```python
numerical_features = x.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()
categorical_features = x.select_dtypes(
    include=["object"]
).columns.tolist()
print("========== NUMERICAL FEATURES ==========")
for feature in numerical_features:
  print("-", feature)

print("\n========== CATEGORICAL FEATURES ==========")
for feature in categorical_features:
  print("-", feature)
```

```python
print("========== CATEGORICAL VALUES ==========")
for column_name in categorical_features:
  print(f"\n{column_name}:")

  print(
      sorted(
          x[column_name].dropna().unique()
      )
  )
```

```python
ordinal_features = [
    "Condition"
]
nominal_features = [
    "Brand",
    "Fuel_Type",
    "Transmission",
    "City",
    "Seller_Type"
]
print("Ordinal features:")
print(ordinal_features)
print("\nNominal features:")
print(nominal_features)
```

```python
def find_iqr_outliers(data, column_name):
  """Find outliers in a numerical column using IQR."""

  q1 = data[column_name].quantile(0.25)
  q3 = data[column_name].quantile(0.75)
  iqr = q3 - q1
  lower_limit = q1 - (1.5 * iqr)
  upper_limit = q3 + (1.5 * iqr)
  outliers = data[
      (data[column_name] < lower_limit)
      | (data[column_name] > upper_limit)
  ]

  return outliers, lower_limit, upper_limit
```

```python
print("========== IQR OUTLIER ANALYSIS ==========")
iqr_summary = []
for column_name in numerical_features:
  outliers, lower_limit, upper_limit = (
      find_iqr_outliers(
          car_data,
          column_name
      )
  )

  iqr_summary.append(
      {
          "Column": column_name,
          "Outlier_Count": len(outliers),
          "Lower_Limit": lower_limit,
          "Upper_Limit": upper_limit
      }
  )

iqr_summary = pd.DataFrame(iqr_summary)
display(iqr_summary)
```

```python
print("========== Z-SCORE OUTLIER ANALYSIS ==========")
z_score_summary = []
for column_name in numerical_features:
  mean_value = car_data[column_name].mean()
  standard_deviation = car_data[
      column_name
  ].std()

  if standard_deviation != 0:
    z_score = (
        car_data[column_name] - mean_value
    ) / standard_deviation

    outlier_count = (
        z_score.abs() > 3
    ).sum()

  else:
    outlier_count = 0

  z_score_summary.append(
      {
          "Column": column_name,
          "Z_Score_Outliers": outlier_count
      }
  )

z_score_summary = pd.DataFrame(
    z_score_summary
)

display(z_score_summary)
```

```python
print("========== OUTLIER METHOD COMPARISION ==========")
outlier_comparision = iqr_summary[
    ["Column", "Outlier_Count"]
].merge(
    z_score_summary,
    on="Column"
)

display(outlier_comparision)
```

```python
car_data_cleaned = car_data.copy()
for column_name in numerical_features:
  q1 = car_data_cleaned[
      column_name
  ].quantile(0.25)
  q3 = car_data_cleaned[
      column_name
  ].quantile(0.75)

  iqr = q3 - q1
  lower_limit = q1 - (1.5 * iqr)
  upper_limit = q3 + (1.5 * iqr)
  car_data_cleaned[column_name] = (
      car_data_cleaned[column_name].clip(
          lower=lower_limit,
          upper=upper_limit
      )
  )

print(
    "Outliers capped using IQR limits"
)
```

```python
print("========== OUTLIERS AFTER IQR CAPPING ==========")
remaining_outliers = []
for column_name in numerical_features:
  outliers, lower_limit, upper_limit = (
      find_iqr_outliers(
          car_data_cleaned,
          column_name
      )
  )

  remaining_outliers.append(
      {
          "Column": column_name,
          "Remaining_Outliers": len(outliers)
      }
  )

remianing_outliers = pd.DataFrame(
    remaining_outliers
)
display(remaining_outliers)
```

```python
x = car_data_cleaned.drop(
    columns=[target_column]
)
y = car_data_cleaned[target_column]
```

```python
x_train, x_test, y_train, y_test = (
    train_test_split(
        x,
        y,
        test_size=0.20,
        random_state=42
    )
)
print("========== TRAIN-TEST SPLIT ==========")
print(f"Training rows: {x_train.shape[0]}")
print(f"Testing rows: {x_test.shape[0]}")
print(f"Training columns: {x_train.shape[1]}")
print(f"Testing columns: {x_test.shape[1]}")
```

```python
one_hot_encoder = OneHotEncoder(
    handle_unknown="ignore",
    sparse_output=False
)
```

```python
condition_order = [
    [
        "Poor",
        "Fair",
        "Good",
        "Very Good",
        "Excellent"
    ]
]

ordinal_encoder = OrdinalEncoder(
    categories=condition_order,
    handle_unknown="use_encoded_value",
    unknown_value=-1
)
```

```python
standard_scaler = StandardScaler()
```

```python
min_max_scaler = MinMaxScaler()
```

```python
standard_preprocessor = ColumnTransformer(
    transformers=[
        (
            "numerical",
            standard_scaler,
            numerical_features
        ),
        (
            "nominal",
            one_hot_encoder,
            nominal_features
        ),
        (
            "ordinal",
            ordinal_encoder,
            ordinal_features
        )
    ],
    remainder="drop"
)
```

```python
x_train_standard = (
    standard_preprocessor.fit_transform(
        x_train
    )
)

x_test_standard = (
    standard_preprocessor.transform(
        x_test
    )
)

print("Standardization fitted only on training data.")
print(
    f"Processed training shape: "
    f"{x_train_standard.shape}"
)
print(
    f"Processed testing shape: "
    f"{x_test_standard.shape}"
)
```

```python
min_max_preprocessor = ColumnTransformer(
    transformers=[
        (
            "numerical",
            min_max_scaler,
            numerical_features
        ),
        (
            "nominal",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            ),
            nominal_features
        ),
        (
            "ordinal",
            OrdinalEncoder(
                categories=condition_order,
                handle_unknown="use_encoded_value",
                unknown_value=-1
            ),
            ordinal_features
        )
    ],
    remainder="drop"
)
```

```python
x_train_min_max = (
    min_max_preprocessor.fit_transform(
        x_train
    )
)

x_test_min_max = (
    min_max_preprocessor.transform(
        x_test
    )
)

print("Min-Max preprocessing completed.")
print(
    f"Min-Max training shape: "
    f"{x_train_min_max.shape}"
)
print(
    f"Min-Max testing shape: "
    f"{x_test_min_max.shape}"
)
```

```python
feature_names = (
    standard_preprocessor
    .get_feature_names_out()
)
print("========== PROCESSED FEATURES ==========")
for feature_name in feature_names:
  print("-", feature_name)
```

```python
x_train_processed = pd.DataFrame(
    x_train_standard,
    columns=feature_names,
    index=x_train.index
)
x_test_processed = pd.DataFrame(
    x_test_standard,
    columns=feature_names,
    index=x_test.index
)

print("Processed DataFrames created successfully.")
```

```python
train_processed = x_train_processed.copy()
train_processed[target_column] = y_train
test_processed = x_test_processed.copy()
test_processed[target_column] = y_test
```

```python
print("========== PROCESSED TRAINING DATA ==========")
display(train_processed.head())
```

```python
print("========== PROCESSED TESTING DATA ==========")
display(test_processed.head())
```

```python
print("========== STADARDIZED NUMERICAL FEATURES ==========")
standardized_columns = [
    column_name
    for column_name in x_train_processed.columns
    if column_name.startswith(
        "numerical__"
    )
]

display(
    x_train_processed[
        standardized_columns
    ].describe()
)
```

```python
min_max_feature_names = (
    min_max_preprocessor
    .get_feature_names_out()
)
min_max_training_data = pd.DataFrame(
    x_train_min_max,
    columns=min_max_feature_names
)
min_max_numerical_columns = [
    column_name
    for column_name in min_max_training_data.columns
    if column_name.startswith(
        "numerical__"
    )
]
print("========== MIN-MAX SCALED FEATURES ==========")
display(
    min_max_training_data[
        min_max_numerical_columns
    ].describe()
)
```

```python
print("========== FINAL MISSING VALUE CHECK ==========")
print(
    "Training missing values:",
    train_processed.isnull().sum().sum()
)
print(
    "Testing missing values:",
    test_processed.isnull().sum().sum()
)
```

```python
print("========== FINAL PROCESSED DATA ==========")
print(
    f"Training dataset shape: "
    f"{train_processed.shape}"
)
print(
    f"Testing dataset shape: "
    f"{test_processed.shape}"
)
```

```python
train_output_file = (
    "Day12_Processed_Used_Car_Train.csv"
)
train_processed.to_csv(
    train_output_file,
    index=False
)
print(
    f"Training dataset saved as "
    f"{train_output_file}"
)
```

```python
test_output_file = (
    "Day_Processed_Used_Car_Test.csv"
)
test_processed.to_csv(
    test_output_file,
    index=False
)
print(
    f"Testing dataset saved as "
    f"{test_output_file}"
)
```

```python
processed_data = pd.concat(
    [
        train_processed,
        test_processed
    ],
    axis=0
)
processed_data = processed_data.sort_index()
print(
    f"Final processed dataset shape: "
    f"{processed_data.shape}"
)
display(processed_data.head())
```
